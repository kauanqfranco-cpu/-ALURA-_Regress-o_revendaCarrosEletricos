<h1> Relação Entre as Variáveis</h1>

Para ilustrar a relação entre as variáveis, já que o trabalho propõe uma inferência causal dos fatores disponibilizados sobre o preço de revenda em dólares dos carros elétricos, faremos um uso de um DAG (Direct Acyclic Graph). 

Um DAG aponta as relações entre as variáveis através do uso de setas. Um bom DAG é muito útil para identificar quais variáveis devem ou não ser incluídas nos modelos para evitar viéses e, também, para ilustrar premissas assumidas durante o estudo.

![DAG](DAG.JPEG)

Por exemplo, de acordo com a imagem acima, se Ano -> Preço de Revenda (USD), implica que o ano de fabricação afeta diretamente o preço de revenda em dólares. É no DAG que também conseguimos observar possíveis viéses de variável omitida e entender melhor como fechar <i><b>backdoor-paths</i></b>.

Para saber mais: https://mixtape.scunning.com/03-directed_acyclical_graphs

```python
def corrMap(df, method = 'pearson'):
    
    import seaborn as sns
    import matplotlib.pyplot as plt
    from matplotlib.colors import LinearSegmentedColormap

    corr = df.corr(numeric_only = True, method = method)
    cList = ['orange', '#EEEEEE', 'green']
    cmap = LinearSegmentedColormap.from_list('orange_to_green', cList)
    fig, ax = plt.subplots(figsize=(11,9))

    sns.heatmap(corr, annot = True, vmin = -1, vmax = 1, cmap = cmap, fmt = '.0%', annot_kws = {'size':8})

    plt.show()
```

<h1>Pré-processamento</h1>

```python
dados = df[['Modelo', 'Valor_revenda(USD)', 'Ano', 'Autonomia(KM)', 'Capacidade_bateria(kWh)']]
X = dados.drop('Valor_revenda(USD)', axis = 1)
y = dados['Valor_revenda(USD)']
```

```python
df.columns
```

```python
xTreino, xTeste, yTreino, yTeste = train_test_split(X, y, random_state = 60494, test_size = 0.2)
```

```python
ct = ColumnTransformer(
    transformers = [('scaler',StandardScaler(), ['Ano', 'Autonomia(KM)', 'Capacidade_bateria(kWh)']),
                   ('encoder', OneHotEncoder(sparse_output = False), ['Modelo'])],
    remainder = 'passthrough')
ct.fit(xTreino)
```

<h1>Treinando Modelos</h1>

XGBoost Regressor
Forest Regressor
Causal Forests

```python
##XGBoost Regressor

xgbPipeline = Pipeline(steps=[
    ('preprocessador', ct),
    ('modelo', xgb.XGBRegressor(objective = 'reg:squarederror', random_state = 64))
])

xgbParamsGrid = {
    'modelo__learning_rate': [0.08, 0.1, 0.12],
    'modelo__gamma': [0, 0.05],
    'modelo__max_depth': [1, 2, 3],
    'modelo__reg_lambda': [4, 5, 6],
    'modelo__reg_alpha': [7, 8, 9],
    'modelo__subsample': [0.6, 0.7, 0.8],  
}

xgbGridCv = GridSearchCV(
    estimator = xgbPipeline,
    param_grid = xgbParamsGrid,
    cv = 5,
    n_jobs = -1,
    scoring = 'neg_mean_squared_error'
)

xgbGridCv.fit(xTreino, yTreino)

xgbPred = xgbGridCv.predict(xTeste)

print(f'R²: {r2_score(yTeste, xgbPred)}')
print(f'MAE: {mean_absolute_error(yTeste, xgbPred)}')
print(f'RMSE: {np.sqrt(mean_squared_error(yTeste, xgbPred))}')
print(f'EV: {explained_variance_score(yTeste, xgbPred)}')
print(50*'-')
print(xgbGridCv.best_params_)
```

```python
##Forest Regressor

RFPipeline = Pipeline(steps=[
    ('preprocessador', ct),
    ('modelo', RandomForestRegressor(n_estimators = 500, random_state = 64, criterion = 'squared_error'))
])

ForestParamsGrid = {
    'modelo__max_depth': [5, 6, 7],
}

ForestGridCv = GridSearchCV(
    estimator = RFPipeline,
    param_grid = ForestParamsGrid,
    cv = 5,
    n_jobs = -1,
    scoring = 'neg_mean_squared_error'
)

ForestGridCv.fit(xTreino, yTreino)

forestPreds = ForestGridCv.predict(xTeste)

print(f'R²: {r2_score(yTeste, forestPreds)}')
print(f'MAE: {mean_absolute_error(yTeste, forestPreds)}')
print(f'RMSE: {np.sqrt(mean_squared_error(yTeste, forestPreds))}')
print(f'EV: {explained_variance_score(yTeste, forestPreds)}')
print(50*'-')
print(ForestGridCv.best_params_)
```

```python
##Ridge-Lasso

elasticPipeline = Pipeline(steps=[
    ('preprocessador', ct),
    ('modelo', ElasticNet(random_state = 64, max_iter = 5000))
])

elasticParamsGrid = {
    'modelo__alpha': [0, 0.01, 0.02],
    'modelo__l1_ratio': [0.6, 0.7, 0.8]
}

elasticCv = GridSearchCV(
    estimator = elasticPipeline,
    param_grid = elasticParamsGrid,
    cv = 5,
    n_jobs = -1,
    scoring = 'neg_mean_squared_error'
)

elasticCv.fit(xTreino, yTreino)

elasticPreds = elasticCv.predict(xTeste)

print(f'R²: {r2_score(yTeste, elasticPreds)}')
print(f'MAE: {mean_absolute_error(yTeste, elasticPreds)}')
print(f'RMSE: {np.sqrt(mean_squared_error(yTeste, elasticPreds))}')
print(f'EV: {explained_variance_score(yTeste, elasticPreds)}')
print(50*'-')
print(elasticCv.best_params_)
```

```python
#SVR

svrPipeline = Pipeline(steps=[
    ('preprocessador', ct),
    ('modelo', SVR(max_iter = 5000))
])

svrParamGrid = {
    'modelo__epsilon': [8, 9, 10, 15, 20, 25, 30, 50, 70, 80],
    'modelo__C': [580, 600, 620],
    'modelo__kernel': ['rbf', 'poly'],
    'modelo__degree': [1, 2, 3],
    'modelo__gamma': ['scale', 'auto']
}

svrGridCv = GridSearchCV(
    estimator = svrPipeline,
    param_grid = svrParamGrid,
    cv = 5,
    n_jobs = -1,
    scoring = 'neg_mean_squared_error'
)

svrGridCv.fit(xTreino, yTreino)

svrPreds = svrGridCv.predict(xTeste)

print(f'R²: {r2_score(yTeste, svrPreds)}')
print(f'MAE: {mean_absolute_error(yTeste, svrPreds)}')
print(f'RMSE: {np.sqrt(mean_squared_error(yTeste, svrPreds))}')
print(f'EV: {explained_variance_score(yTeste, svrPreds)}')
print(50*'-')
print(svrGridCv.best_params_)
```
